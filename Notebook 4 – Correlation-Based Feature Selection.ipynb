```python
# ==========================================================
# Notebook 4 : Correlation-Based Feature Selection
# ==========================================================

import os
import warnings
import numpy as np
import pandas as pd

warnings.filterwarnings("ignore")

print("Libraries Imported Successfully.")
```

```python
# ==========================================================
# Load Numeric Dataset
# ==========================================================

DATA_PATH = "./Datasets/CICIDS2017/processed/cicids2017_numeric.csv"

data = pd.read_csv(
    DATA_PATH,
    low_memory=False
)

print("Dataset Loaded Successfully.")
print(f"Shape: {data.shape}")
```

```python
# ==========================================================
# Separate Features and Labels
# ==========================================================

X = data.drop(columns=["label"])
y = data["label"]

print("=" * 60)
print("Dataset Summary")
print("=" * 60)

print(f"Features Shape : {X.shape}")
print(f"Labels Shape   : {y.shape}")
```

```python
# ==========================================================
# Verify Features
# ==========================================================

print("=" * 60)
print("Feature Verification")
print("=" * 60)

print(f"Total Features : {X.shape[1]}")
print(f"Missing Values : {X.isnull().sum().sum()}")
print(f"Infinite Values: {np.isinf(X.to_numpy()).sum()}")
```

```python
# ==========================================================
# Compute Correlation Matrix
# ==========================================================

print("=" * 60)
print("Computing Correlation Matrix...")
print("=" * 60)

corr_matrix = X.corr().abs()

print("Correlation Matrix Shape:")
print(corr_matrix.shape)
```

```python
# ==========================================================
# Identify Highly Correlated Features
# ==========================================================

CORRELATION_THRESHOLD = 0.95

upper = corr_matrix.where(
    np.triu(
        np.ones(corr_matrix.shape),
        k=1
    ).astype(bool)
)

drop_columns = [
    column
    for column in upper.columns
    if any(upper[column] > CORRELATION_THRESHOLD)
]

print("=" * 60)
print("Correlation-Based Feature Selection")
print("=" * 60)

print(f"Correlation Threshold : {CORRELATION_THRESHOLD}")
print(f"Original Features     : {X.shape[1]}")
print(f"Features Removed      : {len(drop_columns)}")
print(f"Remaining Features    : {X.shape[1] - len(drop_columns)}")
```

```python
# ==========================================================
# Save Removed Features
# ==========================================================

import os

SAVE_PATH = "./Datasets/CICIDS2017/feature_selection"
os.makedirs(SAVE_PATH, exist_ok=True)

removed_features = pd.DataFrame({
    "Removed_Feature": drop_columns
})

REMOVED_FILE = os.path.join(
    SAVE_PATH,
    "removed_features_corr95.csv"
)

removed_features.to_csv(
    REMOVED_FILE,
    index=False
)

print("=" * 60)
print("Removed Features Saved Successfully")
print("=" * 60)
print(f"Location : {REMOVED_FILE}")
print(f"Total Removed : {len(removed_features)}")
```

```python
# ==========================================================
# Create Correlation-Reduced Dataset
# ==========================================================

X_corr = X.drop(columns=drop_columns)

corr_data = X_corr.copy()
corr_data["label"] = y

print("=" * 60)
print("Correlation-Reduced Dataset")
print("=" * 60)

print(f"Features : {X_corr.shape[1]}")
print(f"Dataset Shape : {corr_data.shape}")
```

```python
# ==========================================================
# Save Correlation-Reduced Dataset
# ==========================================================

import os

SAVE_PATH = "./Datasets/CICIDS2017/feature_selection"
os.makedirs(SAVE_PATH, exist_ok=True)

CORR_FILE = os.path.join(
    SAVE_PATH,
    "cicids2017_corr66.csv"
)

corr_data.to_csv(
    CORR_FILE,
    index=False
)

print("=" * 60)
print("Correlation-Reduced Dataset Saved Successfully")
print("=" * 60)
print(f"Location : {CORR_FILE}")
print(f"Shape    : {corr_data.shape}")
```
