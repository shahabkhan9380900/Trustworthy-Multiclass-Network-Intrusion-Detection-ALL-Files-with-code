```python
# ==========================================================
# Notebook 5 : Random Forest Feature Selection
# ==========================================================

import os
import warnings
import numpy as np
import pandas as pd

from sklearn.ensemble import RandomForestClassifier

warnings.filterwarnings("ignore")

print("Libraries Imported Successfully.")
```

```python
# ==========================================================
# Load Correlation-Reduced Dataset
# ==========================================================

DATA_PATH = "./Datasets/CICIDS2017/feature_selection/cicids2017_corr66.csv"

data = pd.read_csv(
    DATA_PATH,
    low_memory=False
)

print("Dataset Loaded Successfully.")
print(f"Shape: {data.shape}")
```

```python
# ==========================================================
# Separate Features and Labels
# ==========================================================

X = data.drop(columns=["label"])
y = data["label"]

print("=" * 60)
print("Dataset Summary")
print("=" * 60)

print(f"Features Shape : {X.shape}")
print(f"Labels Shape   : {y.shape}")
```

```python
# ==========================================================
# Verify Dataset
# ==========================================================

print("=" * 60)
print("Dataset Verification")
print("=" * 60)

print(f"Missing Values : {X.isnull().sum().sum()}")
print(f"Infinite Values: {np.isinf(X.to_numpy()).sum()}")
print(f"Total Features : {X.shape[1]}")
print(f"Total Classes  : {y.nunique()}")
```

```python
# ==========================================================
# Train Random Forest
# ==========================================================

rf = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1,
    bootstrap=True,
    oob_score=False
)

print("=" * 60)
print("Training Random Forest...")
print("=" * 60)

rf.fit(X, y)

print("=" * 60)
print("Training Completed Successfully")
print("=" * 60)
```

```python
# ==========================================================
# Compute Feature Importance
# ==========================================================

importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": rf.feature_importances_
})

importance = importance.sort_values(
    by="Importance",
    ascending=False
).reset_index(drop=True)

print("=" * 60)
print("Top 40 Important Features")
print("=" * 60)

print(importance.head(40))
```

```python
# ==========================================================
# Save Complete Feature Importance
# ==========================================================

SAVE_PATH = "./Datasets/CICIDS2017/feature_selection"
os.makedirs(SAVE_PATH, exist_ok=True)

IMPORTANCE_FILE = os.path.join(
    SAVE_PATH,
    "RF_Feature_Importance.csv"
)

importance.to_csv(
    IMPORTANCE_FILE,
    index=False
)

print("=" * 60)
print("Feature Importance Saved Successfully")
print("=" * 60)
print(f"Location : {IMPORTANCE_FILE}")
print(f"Total Features : {len(importance)}")
```

```python
# ==========================================================
# Extract Top 40 Features
# ==========================================================

TOP_K = 40

top_features = importance.head(TOP_K)["Feature"].tolist()

top40_df = pd.DataFrame({
    "Feature": top_features
})

print("=" * 60)
print("Top 40 Features")
print("=" * 60)

print(top40_df)
```

```python
# ==========================================================
# Save Top 40 Feature List
# ==========================================================

TOP40_FILE = os.path.join(
    SAVE_PATH,
    "Top40_Features.csv"
)

top40_df.to_csv(
    TOP40_FILE,
    index=False
)

print("=" * 60)
print("Top 40 Feature List Saved Successfully")
print("=" * 60)
print(f"Location : {TOP40_FILE}")
```

```python
# ==========================================================
# Create Final 40-Feature Dataset
# ==========================================================

selected_data = data[top_features + ["label"]]

print("=" * 60)
print("Final Selected Dataset")
print("=" * 60)

print(f"Features : {len(top_features)}")
print(f"Dataset Shape : {selected_data.shape}")
```

```python
# ==========================================================
# Save Final 40-Feature Dataset
# ==========================================================

FINAL_FILE = os.path.join(
    SAVE_PATH,
    "cicids2017_selected40.csv"
)

selected_data.to_csv(
    FINAL_FILE,
    index=False
)

print("=" * 60)
print("Final 40-Feature Dataset Saved Successfully")
print("=" * 60)
print(f"Location : {FINAL_FILE}")
print(f"Shape    : {selected_data.shape}")
```
