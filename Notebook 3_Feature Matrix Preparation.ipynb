```python
# ==========================================================
# Notebook 3 : Prepare Feature Matrix
# ==========================================================

import os
import warnings
import numpy as np
import pandas as pd

warnings.filterwarnings("ignore")

pd.set_option("display.max_columns", None)

print("Libraries Imported Successfully.")
```

```python
# ==========================================================
# Load Clean Dataset
# ==========================================================

DATA_PATH = "./Datasets/CICIDS2017/cleaned/cicids2017_clean.csv"

cicids2017 = pd.read_csv(
    DATA_PATH,
    low_memory=False
)

print("Dataset Loaded Successfully.")
print(f"Shape: {cicids2017.shape}")
```

```python
# ==========================================================
# Separate Features and Labels
# ==========================================================

LABEL_COLUMN = "label"

DROP_COLUMNS = [
    "flow_id",
    "timestamp",
    "src_ip",
    "dst_ip",
    "protocol"
]

X = cicids2017.drop(
    columns=DROP_COLUMNS + [LABEL_COLUMN]
)

y = cicids2017[LABEL_COLUMN]

print("=" * 60)
print("Feature Matrix")
print("=" * 60)

print(f"Features Shape : {X.shape}")
print(f"Labels Shape   : {y.shape}")
```

```python
# ==========================================================
# Verify Feature Data Types
# ==========================================================

print("=" * 60)
print("Feature Data Types")
print("=" * 60)

print(X.dtypes.value_counts())

print("\nNon-Numeric Columns Remaining:")

non_numeric = X.select_dtypes(exclude=[np.number]).columns.tolist()

print(non_numeric)

print("\nTotal Features:", X.shape[1])
```

```python
# ==========================================================
# Verify Class Distribution
# ==========================================================

print("=" * 60)
print("Class Distribution")
print("=" * 60)

print(y.value_counts())
```

```python
# ==========================================================
# Save Numeric Dataset
# ==========================================================

import os

SAVE_PATH = "./Datasets/CICIDS2017/processed"
os.makedirs(SAVE_PATH, exist_ok=True)

SAVE_FILE = os.path.join(
    SAVE_PATH,
    "cicids2017_numeric.csv"
)

processed_data = X.copy()
processed_data["label"] = y

processed_data.to_csv(
    SAVE_FILE,
    index=False
)

print("=" * 60)
print("Numeric Dataset Saved Successfully")
print("=" * 60)
print(f"Location : {SAVE_FILE}")
print(f"Shape    : {processed_data.shape}")
```
