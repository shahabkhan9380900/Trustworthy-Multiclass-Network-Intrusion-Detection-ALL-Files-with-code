```python
# ==========================================================
# Notebook 2 : Data Cleaning
# ==========================================================

import os
import warnings

import numpy as np
import pandas as pd

warnings.filterwarnings("ignore")

pd.set_option("display.max_columns", None)

print("Libraries Imported Successfully.")
```

```python
# ==========================================================
# Load Raw Dataset
# ==========================================================

DATA_PATH = "./Datasets/CICIDS2017/raw/cicids2017_raw.csv"

cicids2017 = pd.read_csv(DATA_PATH, low_memory=False)

print("Dataset Loaded Successfully.")

print(f"Shape: {cicids2017.shape}")
```

```python
# ==========================================================
# Verify Dataset Before Cleaning
# ==========================================================

print("=" * 60)
print("Dataset Shape")
print("=" * 60)
print(cicids2017.shape)

print("\n" + "=" * 60)
print("Missing Values")
print("=" * 60)
print(cicids2017.isnull().sum().sum())

print("\n" + "=" * 60)
print("Infinite Values")
print("=" * 60)

numeric_df = cicids2017.select_dtypes(include=[np.number])
print(np.isinf(numeric_df).sum().sum())

print("\n" + "=" * 60)
print("Duplicate Rows")
print("=" * 60)
print(cicids2017.duplicated().sum())
```

```python
# ==========================================================
# Remove Duplicate Rows
# ==========================================================

rows_before = len(cicids2017)

cicids2017 = cicids2017.drop_duplicates().reset_index(drop=True)

rows_after = len(cicids2017)

duplicates_removed = rows_before - rows_after

print("=" * 60)
print("Duplicate Removal Summary")
print("=" * 60)

print(f"Rows Before : {rows_before:,}")
print(f"Rows After  : {rows_after:,}")
print(f"Removed     : {duplicates_removed:,}")
```

```python
# ==========================================================
# Verify Duplicate Removal
# ==========================================================

print("=" * 60)
print("Verification After Duplicate Removal")
print("=" * 60)

print(f"Dataset Shape : {cicids2017.shape}")
print(f"Duplicate Rows: {cicids2017.duplicated().sum()}")
```

```python
# ==========================================================
# Clean Column Names
# ==========================================================

cicids2017.columns = cicids2017.columns.str.strip()

print("Column names cleaned successfully.")

print(f"Total Columns: {len(cicids2017.columns)}")
```

```python
# ==========================================================
# Clean Label Values
# ==========================================================

LABEL_COLUMN = "label"

cicids2017[LABEL_COLUMN] = (
    cicids2017[LABEL_COLUMN]
    .astype(str)
    .str.strip()
)

print("Label values cleaned successfully.")

print("\nUnique Labels:")

print(sorted(cicids2017[LABEL_COLUMN].unique()))
```

```python
# ==========================================================
# Verify Data Types
# ==========================================================

print("=" * 60)
print("Data Types")
print("=" * 60)

print(cicids2017.dtypes.value_counts())

print("\nObject Columns:")

print(cicids2017.select_dtypes(include=["object"]).columns.tolist())
```

```python
# ==========================================================
# Save Clean Dataset
# ==========================================================

import os

SAVE_PATH = "./Datasets/CICIDS2017/cleaned"
os.makedirs(SAVE_PATH, exist_ok=True)

SAVE_FILE = os.path.join(SAVE_PATH, "cicids2017_clean.csv")

cicids2017.to_csv(
    SAVE_FILE,
    index=False
)

print("=" * 60)
print("Clean Dataset Saved Successfully")
print("=" * 60)
print(f"Location : {SAVE_FILE}")
print(f"Shape    : {cicids2017.shape}")
```
