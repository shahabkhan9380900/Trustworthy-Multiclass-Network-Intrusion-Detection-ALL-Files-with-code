```python
# ==========================================================
# Notebook 1: Load Original CICIDS2017 Dataset
# ==========================================================

import os
import glob
import warnings

import numpy as np
import pandas as pd

warnings.filterwarnings("ignore")

# Display settings
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 10)

print("Libraries Imported Successfully.")
```

```python
# ==========================================================
# Set Dataset Path
# ==========================================================

DATASET_PATH = r".\Datasets\CICIDS2017/orignal dataset"

csv_files = sorted(glob.glob(os.path.join(DATASET_PATH, "*.csv")))

print(f"Number of CSV Files Found: {len(csv_files)}")

for file in csv_files:
    print(os.path.basename(file))
```

```python
# ==========================================================
# Load All CSV Files
# ==========================================================

dataframes = []

for file in csv_files:

    print(f"Loading: {os.path.basename(file)}")

    df = pd.read_csv(file, low_memory=False)

    print(f"Shape: {df.shape}")

    dataframes.append(df)

print("\nAll CSV files loaded successfully.")
print(f"Total DataFrames Loaded: {len(dataframes)}")
```

```python
# ==========================================================
# Verify Column Consistency Across All CSV Files
# ==========================================================

reference_columns = dataframes[0].columns.tolist()

all_match = True

for i, df in enumerate(dataframes):
    if df.columns.tolist() != reference_columns:
        print(f"❌ Column mismatch found in: {os.path.basename(csv_files[i])}")
        all_match = False

if all_match:
    print("✅ All CSV files have identical column names and column order.")
```

```python
# ==========================================================
# Merge All CSV Files
# ==========================================================

cicids2017_raw = pd.concat(dataframes, ignore_index=True)

print("=" * 60)
print("Merged Dataset Created Successfully")
print("=" * 60)

print(f"Dataset Shape: {cicids2017_raw.shape}")
```

```python
# ==========================================================
# Basic Dataset Information
# ==========================================================

print(cicids2017_raw.info())
```

```python
# ==========================================================
# Check Object Columns
# ==========================================================

object_columns = cicids2017_raw.select_dtypes(include=['object']).columns

print(f"Number of Object Columns: {len(object_columns)}\n")

for col in object_columns:
    print(col)
```

```python
# ==========================================================
# Dataset Summary
# ==========================================================

print("=" * 60)
print("Dataset Shape")
print("=" * 60)
print(cicids2017_raw.shape)

print("\n" + "=" * 60)
print("Missing Values")
print("=" * 60)
print(cicids2017_raw.isnull().sum().sum())

print("\n" + "=" * 60)
print("Infinite Values")
print("=" * 60)

numeric_df = cicids2017_raw.select_dtypes(include=[np.number])

print(np.isinf(numeric_df).sum().sum())

print("\n" + "=" * 60)
print("Duplicate Rows")
print("=" * 60)

print(cicids2017_raw.duplicated().sum())
```

```python
# ==========================================================
# Save Merged Raw Dataset
# ==========================================================

OUTPUT_PATH = "./Datasets/CICIDS2017/raw"

import os
os.makedirs(OUTPUT_PATH, exist_ok=True)

cicids2017_raw.to_csv(
    os.path.join(OUTPUT_PATH, "cicids2017_raw.csv"),
    index=False
)

print("Raw merged dataset saved successfully.")
```
